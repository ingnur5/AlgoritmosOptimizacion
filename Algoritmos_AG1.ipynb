##Devolución de cambio por técnica voraz

```
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## 1. Primero en 1D

Si los números están en una sola línea,los ordeno y miro cuáles quedan uno al lado del otro. Ese es es la respuesta,si dos números no quedan de vecinos después de ordenar, es porque hay otro número metido entre ellos, y ese número de en medio va a estar más cerca de uno de los dos que ellos entre sí. Así que
el par más cercano nunca se salta vecinos.


```
import math
import random
import time
import matplotlib.pyplot as plt

def par_mas_cercano_1d(numeros):
    # ordeno la lista y comparo numeros que quedan seguidos
    if len(numeros) < 2:
        return None

    ordenados = sorted(numeros)
    mejor_distancia = float('inf')
    mejor_par = None

    for i in range(len(ordenados) - 1):
        distancia_actual = abs(ordenados[i + 1] - ordenados[i])
        if distancia_actual < mejor_distancia:
            mejor_distancia = distancia_actual
            mejor_par = (ordenados[i], ordenados[i + 1])

    return mejor_par, mejor_distancia
```

```
numeros = [3403, 4537, 9089, 9746, 7259, 1204, 6321, 8890, 2211]
par, distancia_encontrada = par_mas_cercano_1d(numeros)
print("numeros:", numeros)
print("par mas cercano:", par, " distancia:", distancia_encontrada)
```

## 2. Fuerza bruta

Se debe comparar cada punto contra todos los demás, saco la distancia entre ellos y me quedo con el par que dio la distancia más peqeña.Sirve igual para 2D, 3D, porque la distancia se calcula sumando las diferencias al cuadrado de cada coordenada.

```
def distancia(punto1, punto2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(punto1, punto2)))


def par_mas_cercano_fuerza_bruta(puntos):
    # comparo cada punto contra todos los demas
    cantidad = len(puntos)
    mejor_distancia = float('inf')
    mejor_par = None

    for i in range(cantidad):
        for j in range(i + 1, cantidad):
            distancia_actual = distancia(puntos[i], puntos[j])
            if distancia_actual < mejor_distancia:
                mejor_distancia = distancia_actual
                mejor_par = (puntos[i], puntos[j])

    return mejor_par, mejor_distancia
```

## 3. Divide y vencerás

Fuerza bruta funciona pero es lenta. Con divide y vencerás lo que hago es
partir el problema en pedazos chicos, resolverlos por separado y después
juntar los resultados con cuidado.

```
def _caso_base(puntos):
    # para grupos chiquitos (3 puntos o menos) no vale la pena complicarse
    mejor_distancia = float('inf')
    mejor_par = None
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            distancia_actual = distancia(puntos[i], puntos[j])
            if distancia_actual < mejor_distancia:
                mejor_distancia = distancia_actual
                mejor_par = (puntos[i], puntos[j])
    return mejor_par, mejor_distancia


def _revisar_franja(franja, mejor_distancia, mejor_par):

    franja.sort(key=lambda punto: punto[1])
    cantidad = len(franja)

    for i in range(cantidad):
        j = i + 1
        while j < cantidad and (franja[j][1] - franja[i][1]) < mejor_distancia and (j - i) <= 7:
            distancia_actual = distancia(franja[i], franja[j])
            if distancia_actual < mejor_distancia:
                mejor_distancia = distancia_actual
                mejor_par = (franja[i], franja[j])
            j += 1

    return mejor_par, mejor_distancia


def _buscar_recursivo(puntos_por_x, puntos_por_y):
    # puntos_por_x va ordenado por x, puntos_por_y por y
    # los dejo ya ordenados para no repetir ese trabajo en cada llamada
    cantidad = len(puntos_por_x)

    if cantidad <= 3:
        return _caso_base(puntos_por_x)

    medio = cantidad // 2
    punto_medio = puntos_por_x[medio]

    izquierda_x = puntos_por_x[:medio]
    derecha_x = puntos_por_x[medio:]

    ids_izquierda = set(map(id, izquierda_x))
    izquierda_y = [punto for punto in puntos_por_y if id(punto) in ids_izquierda]
    derecha_y = [punto for punto in puntos_por_y if id(punto) not in ids_izquierda]

    par_izquierda, distancia_izquierda = _buscar_recursivo(izquierda_x, izquierda_y)
    par_derecha, distancia_derecha = _buscar_recursivo(derecha_x, derecha_y)

    if distancia_izquierda <= distancia_derecha:
        mejor_par, mejor_distancia = par_izquierda, distancia_izquierda
    else:
        mejor_par, mejor_distancia = par_derecha, distancia_derecha

    franja = [punto for punto in puntos_por_y if abs(punto[0] - punto_medio[0]) < mejor_distancia]
    mejor_par, mejor_distancia = _revisar_franja(franja, mejor_distancia, mejor_par)

    return mejor_par, mejor_distancia


def par_mas_cercano_divide_venceras_2d(puntos):
    if len(puntos) < 2:
        return None

    puntos_por_x = sorted(puntos, key=lambda punto: punto[0])
    puntos_por_y = sorted(puntos, key=lambda punto: punto[1])
    return _buscar_recursivo(puntos_por_x, puntos_por_y)
```

## 5. Ahora en 3D

La lógica es casi la misma que en 2D, solo que ahora cada punto tiene una
coordenada más (`x, y, z`) y la distancia se calcula con las tres. La diferencia está en la franja de en medio: en 2D bastaba revisar unos 7
vecinos, pero en 3D hay más espacio para que quepan puntos cerca sin pasarse
de la distancia mínima. Por eso reviso más vecinos, uso 40, para no dejar
pasar por error el par correcto.


```
def _revisar_franja_3d(franja, mejor_distancia, mejor_par):
    franja.sort(key=lambda punto: punto[1])
    cantidad = len(franja)
    limite_vecinos = 40

    for i in range(cantidad):
        j = i + 1
        while j < cantidad and (franja[j][1] - franja[i][1]) < mejor_distancia and (j - i) <= limite_vecinos:
            distancia_actual = distancia(franja[i], franja[j])
            if distancia_actual < mejor_distancia:
                mejor_distancia = distancia_actual
                mejor_par = (franja[i], franja[j])
            j += 1

    return mejor_par, mejor_distancia


def _buscar_recursivo_3d(puntos_por_x, puntos_por_y):
    cantidad = len(puntos_por_x)

    if cantidad <= 3:
        return _caso_base(puntos_por_x)

    medio = cantidad // 2
    punto_medio = puntos_por_x[medio]

    izquierda_x = puntos_por_x[:medio]
    derecha_x = puntos_por_x[medio:]

    ids_izquierda = set(map(id, izquierda_x))
    izquierda_y = [punto for punto in puntos_por_y if id(punto) in ids_izquierda]
    derecha_y = [punto for punto in puntos_por_y if id(punto) not in ids_izquierda]

    par_izquierda, distancia_izquierda = _buscar_recursivo_3d(izquierda_x, izquierda_y)
    par_derecha, distancia_derecha = _buscar_recursivo_3d(derecha_x, derecha_y)

    if distancia_izquierda <= distancia_derecha:
        mejor_par, mejor_distancia = par_izquierda, distancia_izquierda
    else:
        mejor_par, mejor_distancia = par_derecha, distancia_derecha

    franja = [punto for punto in puntos_por_y if abs(punto[0] - punto_medio[0]) < mejor_distancia]
    mejor_par, mejor_distancia = _revisar_franja_3d(franja, mejor_distancia, mejor_par)

    return mejor_par, mejor_distancia


def par_mas_cercano_divide_venceras_3d(puntos):
    if len(puntos) < 2:
        return None

    puntos_por_x = sorted(puntos, key=lambda punto: punto[0])
    puntos_por_y = sorted(puntos, key=lambda punto: punto[1])
    return _buscar_recursivo_3d(puntos_por_x, puntos_por_y)
```

```
puntos_3d = [(1122, 6175, 300), (135, 4076, 900), (7296, 2741, 120),
             (500, 4100, 890), (7300, 2750, 125), (3000, 3000, 3000),
             (1200, 6200, 305), (9999, 1, 50)]

par_bruto_3d, distancia_bruta_3d = par_mas_cercano_fuerza_bruta(puntos_3d)
print("fuerza bruta ->", par_bruto_3d, " distancia:", round(distancia_bruta_3d, 4))

par_dyv_3d, distancia_dyv_3d = par_mas_cercano_divide_venceras_3d(puntos_3d)
print("divide y venceras ->", par_dyv_3d, " distancia:", round(distancia_dyv_3d, 4))

assert round(distancia_bruta_3d, 6) == round(distancia_dyv_3d, 6)
print("dan lo mismo, todo bien")
```

Divide y vencerá evidentemente es más rápido que el algoritmo de fuerza bruta

```
figura, eje = plt.subplots(figsize=(6, 5))
metodos = ["Fuerza bruta\nO(n^2)", "Divide y Venceras\nO(n log n)"]
tiempos = [tiempo_bruto, tiempo_dyv]
colores = ["#C44E52", "#55A868"]

barras = eje.bar(metodos, tiempos, color=colores)
eje.set_ylabel("segundos")
eje.set_title(f"Tiempo de ejecucion con {len(puntos_al_azar)} puntos")

for barra, tiempo in zip(barras, tiempos):
    eje.annotate(f"{tiempo:.3f}s", (barra.get_x() + barra.get_width()/2, barra.get_height()),
                 textcoords="offset points", xytext=(0, 5), ha='center')

plt.tight_layout()
plt.show()
```

| Método | Qué tan rápido es | Por qué |
| :--- | :--- | :--- |
| **1D** (ordenar y ver vecinos) | $O(n \log n)$ | Lo único que demora es ordenar. |
| **Fuerza bruta** (2D/3D) | $O(n^2)$ | Comparo todo contra todo, por eso se pone lento. |
| **Divide y vencerás 2D** | $O(n \log n)$ | Parto el problema a la mitad y juntar los resultados es rápido (7 vecinos por punto en la franja). |
| **Divide y vencerás 3D** | $O(n \log n)$ | Lo mismo, solo que en la franja reviso más vecinos por el espacio extra. |
